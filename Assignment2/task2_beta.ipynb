```python
import pandas as pd
import numpy as np
import yfinance as yf

print("Libraries imported successfully!")
```

```python
tickers = ["PG", "SPY"]

prices = yf.download(
    tickers,
    start="2021-09-01",
    end="2026-10-01",
    auto_adjust=True,
    progress=False
)["Close"]

print(prices.tail())
```

```python
monthly_prices = prices.resample("ME").last()
monthly_returns = monthly_prices.pct_change(fill_method=None).dropna()

beta = monthly_returns["PG"].cov(monthly_returns["SPY"]) / monthly_returns["SPY"].var()

print("Number of monthly returns:", len(monthly_returns))
print("First return month:", monthly_returns.index.min().date())
print("Last return month:", monthly_returns.index.max().date())
print(f"P&G Beta: {beta:.4f}")
```
