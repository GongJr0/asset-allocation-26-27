## Report

**Report - Core Part**

We used daily adjusted closing prices of SPY from Yahoo Finance from its launch in January 1993 to November 2026. The first 12 months only build the first signal, so both strategies are evaluated from February 1994 to November 2026 (about 33 years). At the last trading day of each month we compute the trailing return

$$R_t^{(L)} = \frac{P_t}{P_{t-L\ \text{months}}} - 1, \qquad L = 12$$

and hold SPY for the next month if $R_t^{(L)} > 0$, otherwise cash (0% return). The signal from the month-end close is applied from the next trading day, so the position on day *t* only uses prices through *t−1*. Every change in position costs 5 bps per side; buy-and-hold pays the entry cost once. Sharpe ratios use a zero risk-free rate.

| Strategy | Ann. return | Volatility | Sharpe | Max drawdown | Time in market | Trades (sides) | Trades / year | Total cost |
|---|---|---|---|---|---|---|---|---|
| Buy & hold | 10.6% | 18.7% | 0.63 | -55.2% | 100% | 1 | – | 0.0% |
| TSMOM 12m | 10.8% | 14.7% | 0.77 | -33.7% | 82% | 17 | 0.5 | 0.8% |

The 12-month TSMOM rule earned slightly more than buy-and-hold (10.8% vs 10.6% per year) with clearly lower volatility (14.7% vs 18.7%). Its Sharpe ratio is 0.77 vs 0.63, so on a risk-adjusted basis it **beats** buy-and-hold. The biggest gain is in downside protection: the maximum drawdown falls from -55% to -34%.

| Flat from | Back in | Months | SPY return missed |
|---|---|---|---|
| Dec 2000 | Jun 2003 | 31 | **-23.2%** |
| Feb 2008 | Oct 2009 | 21 | **-21.4%** |
| Jun 2012 | Jun 2012 | 1 | +4.1% |
| Oct 2015 | Oct 2015 | 1 | +8.5% |
| Feb 2016 | Mar 2016 | 2 | +6.6% |
| Jan 2019 | Feb 2019 | 2 | +11.5% |
| Apr 2020 | Apr 2020 | 1 | +12.7% |
| Jun 2022 | Apr 2023 | 11 | +2.4% |

The whole result comes from **two long bear markets**. The rule was out for 31 months in 2000–2003 and 21 months in 2008–2009, and SPY lost 23% and 21% while it was flat. These two episodes create the gap between the equity curves, which opens in 2001 and widens in 2008.

The other six flat periods are **whipsaws**. After a sharp but short sell-off (2015–16, Q4 2018, March 2020) the 12-month return turned negative just as the market bottomed. The rule then sold near the low and missed a fast rebound of +4% to +13%, and SPY was higher each time the rule bought back. Because the signal reacts slowly, these losses add up: from 2016 the lead over buy-and-hold shrinks, and at the end of the sample TSMOM is only slightly ahead.

The worst case was 2020. The 12-month return was still positive at the end of February, so the rule stayed fully invested through the COVID crash. That is why its maximum drawdown (-34%) is about as deep as buy-and-hold's in 2020. It then sold at the end of March and missed April's +12.7%. The rule can protect against long bear markets but not against fast crashes. In 2022 it went flat only in June, after much of the fall had already happened, so it avoided the second leg of the decline but gained little overall (+2.4% missed).

Turnover is very low: 17 trades (8 exits and re-entries) over 33 years, about 0.5 per year, and the total cost paid is only 0.8%. Transaction costs are negligible for this rule; even at 10× the cost the result would barely change. Note also that cash is assumed to earn 0%. With T-bill interest on the 18% of days spent in cash, TSMOM's return would be somewhat higher.

**Report - Stretch Part**

We repeated the backtest for formation windows of 3 to 12 months and split the sample in half on 23 June 2010. We selected the window with the highest Sharpe on the first half only and evaluated it once on the second half.

| Window | Sharpe (full) | Sharpe 1st half (IS) | Sharpe 2nd half (OOS) | Trades / year |
|---|---|---|---|---|
| 3m | 0.64 | 0.46 | 0.84 | 2.5 |
| 4m | 0.70 | 0.52 | **0.90** | 2.3 |
| 5m | 0.72 | 0.71 | 0.73 | 1.6 |
| **6m (selected)** | 0.71 | **0.82** | 0.61 | 1.4 |
| 7m | 0.71 | 0.75 | 0.68 | 1.1 |
| 8m | 0.73 | 0.72 | 0.73 | 1.1 |
| 9m | 0.71 | 0.81 | 0.63 | 1.0 |
| 10m | 0.80 | 0.81 | 0.79 | 0.8 |
| 11m | 0.72 | 0.72 | 0.73 | 0.8 |
| 12m | 0.77 | 0.80 | 0.76 | 0.5 |
| *Average of 10 windows* | | *0.71* | *0.74* | |
| *Buy & hold* | *0.63* | *0.44* | *0.87* | – |

The best window on the first half is **6 months**, with an IS Sharpe of **0.82**. Evaluated once on the second half, its Sharpe falls to **0.61**, which is **0.21 lower (-26%)**. The selected window ranks **last of 10** out of sample. The best OOS window would have been 4 months (0.90), which ranked 9th of 10 in sample. The standard 12-month window, which needs no optimisation, does better OOS (0.76) than the optimised 6-month window.

The fall is not caused by a weaker market: buy-and-hold's Sharpe *doubled* from 0.44 to 0.87, and the average window rose from 0.71 to 0.74. The cleaner comparison is against the average window. In sample, the 6-month window beat the average by +0.11. Out of sample it was 0.13 *below* the average. The in-sample ranking of the windows does not just fail to persist, it reverses: the Spearman rank correlation between IS and OOS Sharpe is **-0.66**. The short windows (3–4 months) that looked worst in 1994–2010 were the best in 2010–2026, and the reverse holds for 6 and 9 months.

Two further points stand out. First, in sample every window beats buy-and-hold (0.46–0.82 vs 0.44), but out of sample only the 4-month window does (0.90 vs 0.87). The value of TSMOM on SPY comes mainly from the two long bear markets in the first half. The second half had few long declines and several V-shaped crashes, where trend-following adds little. Second, the windows differ in Sharpe by only about ±0.1, which is small compared with the noise in a Sharpe estimate over 16 years. The "best" window is therefore mostly chosen by noise.

**Link to López de Prado.** This is the selection bias that López de Prado calls *backtest overfitting*. When several configurations are tried on the same data and the best one is reported, its in-sample Sharpe is biased upward even if no configuration is better than the others. By the False Strategy Theorem, the expected maximum Sharpe of $N$ trials with no true skill is

$$E[\max SR] \approx \sqrt{V[\widehat{SR}]}\left((1-\gamma)\,\Phi^{-1}\!\left(1-\tfrac{1}{N}\right) + \gamma\,\Phi^{-1}\!\left(1-\tfrac{1}{Ne}\right)\right)$$

where $\gamma$ is the Euler–Mascheroni constant and $V[\widehat{SR}]$ is the variance of the Sharpe ratios across the trials. The Deflated Sharpe Ratio (Bailey & López de Prado, 2014) is the probability that the true Sharpe is above this benchmark, corrected for skewness and fat tails.

| Measure (first half, best window = 6m) | Value |
|---|---|
| Expected max Sharpe of 10 unskilled trials (ann.) | 0.20 |
| In-sample Sharpe of the best window (ann.) | 0.82 |
| Probabilistic Sharpe Ratio (vs 0) | 1.00 |
| Deflated Sharpe Ratio (vs expected max) | 0.99 |

The DSR of 0.99 is above the 0.95 threshold, so the 6-month Sharpe of 0.82 is significant even after correcting for the 10 trials. This does **not** contradict the out-of-sample result, because the two tests answer different questions:

- **The DSR asks whether the strategy family has a positive Sharpe.** It does: all 10 windows have Sharpe ratios around 0.7 in sample, and they stay around 0.7 out of sample (average 0.74). Trend-following on SPY carried a real return over this period.
- **The out-of-sample test asks whether *picking the best window* adds value.** It does not. The windows are highly correlated, so the variance of their Sharpe ratios is small and the benchmark $E[\max SR]$ is only 0.20. The differences *between* windows (±0.1) are noise, and the in-sample winner turned out to be the out-of-sample loser.

The single split is one draw of López de Prado's *Probability of Backtest Overfitting* (PBO; Bailey, Borwein, López de Prado & Zhu, 2017), which measures how often the in-sample best configuration ends up below the out-of-sample median. Here it ends up last, the clearest sign of overfitting. This is why López de Prado recommends reporting the number of trials and deflating the result, rather than reporting only the best backtest.

**Conclusion.** On SPY over 1994–2026, 12-month time-series momentum beats buy-and-hold on a risk-adjusted basis: Sharpe 0.77 vs 0.63, and maximum drawdown -34% vs -55%, at a total cost of under 1%. The whole edge comes from sitting out the 2000–03 and 2008–09 bear markets; in sharp V-shaped sell-offs (2015–16, 2018, 2020) the rule sells near the bottom and loses. Optimising the formation window does not help. The window with the best in-sample Sharpe (6 months) falls from 0.82 to 0.61 out of sample and ranks last of 10, as the selection-bias argument of López de Prado predicts. The robust finding is that trend-following as a whole works over this sample, not that any particular window is better; a standard, pre-specified 12-month window is the safer choice.


## 0.Set up

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from scipy import stats as st

TICKER = "SPY"
START = "1993-01-01"           # SPY inception; full history
LOOKBACK = 12                    # core formation window, months
GRID = list(range(3, 13))        # stretch: 3..12 months
COST_BPS = 5                     # per side
COST = COST_BPS / 1e4
```

## 1. Data


```python
def load_prices(ticker, start):
    try:
        import yfinance as yf
        d = yf.download(ticker, start=start, auto_adjust=True, progress=False)
        px = d["Close"].squeeze().dropna()
        if len(px) > 1000:
            return px.rename(ticker)
    except Exception as e:
        print(f"yfinance failed ({e}); trying Stooq")
    url = f"https://stooq.com/q/d/l/?s={ticker.lower()}.us&i=d"
    d = pd.read_csv(url, index_col="Date", parse_dates=True)
    return d["Close"].loc[start:].rename(ticker)


px = load_prices(TICKER, START)
ret = px.pct_change()

# last trading day of each month = rebalance dates
month_end = px.groupby(px.index.to_period("M")).tail(1)
month_end = month_end.iloc[:-1] if month_end.index[-1] == px.index[-1] else month_end  # drop unfinished month

# common sample: start once the longest window has a signal, so every strategy covers the same days
start_day = px.index[px.index.get_loc(month_end.index[max(GRID)]) + 1]   # first day with a 12m signal
print(f"{TICKER}: {px.index[0]:%Y-%m-%d} to {px.index[-1]:%Y-%m-%d}; "
      f"evaluation from {start_day:%Y-%m-%d} ({len(px.loc[start_day:])} days, "
      f"{len(px.loc[start_day:]) / 252:.1f} years)")
```

## 2.Strategy

```python
def tsmom_position(lookback):
    """Daily position: sign of the trailing `lookback`-month return at each month-end, long/flat.
    The signal from the close of month-end m is held from the next trading day to the next
    month-end, i.e. the position on day t only uses prices through t-1."""
    signal = (month_end / month_end.shift(lookback) - 1 > 0).astype(float)
    signal[month_end.shift(lookback).isna()] = np.nan
    return signal.reindex(px.index).ffill().shift(1)      # shift(1): trade the day after the signal


def backtest(pos, start=start_day, end=None):
    held = pos.loc[start:end].fillna(0.0)
    turnover = held.diff().abs()
    turnover.iloc[0] = abs(held.iloc[0])                    # entry cost if the test starts long
    net = held * ret.loc[start:end] - turnover * COST
    return pd.DataFrame({"net": net, "pos": held, "turnover": turnover})


def buy_hold(start=start_day, end=None):
    r = ret.loc[start:end].copy()
    r.iloc[0] = -COST                                       # bought at the close of day 0
    turnover = pd.Series(0.0, index=r.index)
    turnover.iloc[0] = 1.0
    return pd.DataFrame({"net": r, "pos": 1.0, "turnover": turnover})


def stats(bt):
    r = bt["net"]
    years = len(r) / 252
    eq = (1 + r).cumprod()
    vol = r.std() * np.sqrt(252)
    return pd.Series({
        "Ann. return (CAGR)": eq.iloc[-1] ** (1 / years) - 1,
        "Ann. volatility": vol,
        "Sharpe (rf=0)": r.mean() * 252 / vol,
        "Max drawdown": (eq / eq.cummax() - 1).min(),
        "Time in market": (bt["pos"] != 0).mean(),
        "Trades (sides)": bt["turnover"].sum(),
        "Trades per year": bt["turnover"].sum() / years,
        "Total cost paid": bt["turnover"].sum() * COST,
    })
```

## 3.Results

```python
pos_core = tsmom_position(LOOKBACK)
results = {"Buy & hold": buy_hold(), f"TSMOM {LOOKBACK}m": backtest(pos_core)}
table = pd.DataFrame({k: stats(v) for k, v in results.items()})
print(f"=== {TICKER}, TSMOM {LOOKBACK}m, long/flat, monthly rebalance, {COST_BPS} bps/side ===")
print(table.round(3).to_string())

# when was the strategy out of the market?
flat = (results[f"TSMOM {LOOKBACK}m"]["pos"] == 0).astype(int)
starts = flat.index[(flat == 1) & (flat.shift(1, fill_value=0) == 0)]
ends = flat.index[(flat == 1) & (flat.shift(-1, fill_value=0) == 0)]
spells = pd.DataFrame({"flat from": starts, "back in": ends})
spells["months"] = ((spells["back in"] - spells["flat from"]).dt.days / 30.4).round(1)
spells[f"{TICKER} return missed"] = [(1 + ret.loc[a:b]).prod() - 1
                                     for a, b in zip(spells["flat from"], spells["back in"])]
print("\nPeriods out of the market (return of the asset on the days the strategy was flat):")
print(spells.round({"months": 1, f"{TICKER} return missed": 3}).to_string(index=False))

fig, axes = plt.subplots(2, 1, figsize=(11, 7), sharex=True, gridspec_kw={"height_ratios": [3, 1]})
for (name, bt), c in zip(results.items(), ["black", "#c44e52"]):
    eq = (1 + bt["net"]).cumprod()
    axes[0].plot(eq.index, eq, label=name, color=c, lw=1.2)
    axes[1].plot(eq.index, eq / eq.cummax() - 1, color=c, lw=0.9)
for a, b in zip(starts, ends):
    axes[0].axvspan(a, b, color="grey", alpha=0.2, lw=0)
axes[0].set_yscale("log")
axes[0].set_title(f"{TICKER}: growth of $1, net of {COST_BPS} bps/side (grey = TSMOM flat)")
axes[0].legend()
axes[1].set_title("Drawdown")
plt.tight_layout()
plt.show()
```

## 4. Stretch: window grid, in-sample vs out-of-sample

```python
mid = px.loc[start_day:].index[len(px.loc[start_day:]) // 2]
rows = {}
for L in GRID:
    p = tsmom_position(L)
    rows[L] = {"Sharpe full": stats(backtest(p))["Sharpe (rf=0)"],
               "Sharpe 1st half (IS)": stats(backtest(p, end=mid))["Sharpe (rf=0)"],
               "Sharpe 2nd half (OOS)": stats(backtest(p, start=mid))["Sharpe (rf=0)"],
               "Trades per year": stats(backtest(p))["Trades per year"]}
grid = pd.DataFrame(rows).T
grid.index.name = "window (months)"

bh_is = stats(buy_hold(end=mid))["Sharpe (rf=0)"]
bh_oos = stats(buy_hold(start=mid))["Sharpe (rf=0)"]
best = grid["Sharpe 1st half (IS)"].idxmax()
is_best = grid.loc[best, "Sharpe 1st half (IS)"]
oos = grid.loc[best, "Sharpe 2nd half (OOS)"]
oos_rank = grid["Sharpe 2nd half (OOS)"].rank(ascending=False)[best]

print(f"=== Window grid, split at {mid:%Y-%m-%d} ===")
print(grid.round(3).to_string())
print(f"\nBuy & hold Sharpe: IS {bh_is:.3f}, OOS {bh_oos:.3f}")
print(f"Best window on 1st half: {best} months, IS Sharpe {is_best:.3f}")
print(f"Same window on 2nd half: OOS Sharpe {oos:.3f}  ->  OOS minus IS = {oos - is_best:+.3f}; "
      f"rank among windows: IS 1, OOS {oos_rank:.0f} of {len(GRID)}")
print(f"Best possible OOS window: {grid['Sharpe 2nd half (OOS)'].idxmax()} months "
      f"(Sharpe {grid['Sharpe 2nd half (OOS)'].max():.3f})")
print(f"Average IS Sharpe across all windows: {grid['Sharpe 1st half (IS)'].mean():.3f}; "
      f"average OOS: {grid['Sharpe 2nd half (OOS)'].mean():.3f}")
```

## 5. Stretch: Deflated Sharpe Ratio (Bailey & López de Prado, 2014)

```python
def prob_sharpe(r, sr_bench):
    """P(true Sharpe > sr_bench) given skew/kurtosis of r. Per-period (daily) Sharpe units."""
    r = r.dropna()
    sr = r.mean() / r.std()
    skew, kurt = st.skew(r), st.kurtosis(r, fisher=False)
    z = (sr - sr_bench) * np.sqrt(len(r) - 1) / np.sqrt(1 - skew * sr + (kurt - 1) / 4 * sr ** 2)
    return st.norm.cdf(z)


def expected_max_sharpe(sr_trials):
    """Expected best Sharpe among len(sr_trials) unskilled trials (False Strategy Theorem)."""
    n, g = len(sr_trials), 0.5772156649                      # g = Euler-Mascheroni constant
    return np.std(sr_trials, ddof=1) * ((1 - g) * st.norm.ppf(1 - 1 / n)
                                        + g * st.norm.ppf(1 - 1 / (n * np.e)))


is_daily = {L: backtest(tsmom_position(L), end=mid)["net"] for L in GRID}
sr0 = expected_max_sharpe([r.mean() / r.std() for r in is_daily.values()])
sr0_ann = sr0 * np.sqrt(252)
psr = prob_sharpe(is_daily[best], 0.0)      # Probabilistic Sharpe Ratio, benchmark 0
dsr = prob_sharpe(is_daily[best], sr0)      # Deflated Sharpe Ratio, benchmark = expected max
print(f"=== Selection bias, {len(GRID)} windows tried ===")
print(f"Expected best Sharpe from {len(GRID)} trials with no skill (ann.): {sr0_ann:.3f}")
print(f"Probabilistic Sharpe Ratio of best window (vs SR=0): {psr:.3f}")
print(f"Deflated Sharpe Ratio of best window (vs SR0):       {dsr:.3f}")

fig, ax = plt.subplots(figsize=(11, 4))
x = np.arange(len(GRID))
ax.bar(x - 0.2, grid["Sharpe 1st half (IS)"], 0.4, label="1st half (in-sample)", color="#4c72b0")
ax.bar(x + 0.2, grid["Sharpe 2nd half (OOS)"], 0.4, label="2nd half (out-of-sample)", color="#dd8452")
ax.axhline(bh_is, color="#4c72b0", ls="--", lw=0.8, label="Buy & hold IS")
ax.axhline(bh_oos, color="#dd8452", ls="--", lw=0.8, label="Buy & hold OOS")
ax.set_xticks(x, [f"{L}m" for L in GRID])
ax.set_title(f"Sharpe ratio by formation window (best IS window = {best}m)")
ax.legend(ncol=2)
plt.tight_layout()
plt.show()
```
